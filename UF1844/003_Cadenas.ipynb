# Cadenas de texto

```python
cadena = """ este es un texto largo
-punto 1: 'Comillas simples'
-punto 2: \"Comillas dobles\"
-punto 3 
"""

print(len(cadena))
```

```python
dato = input("Dame un numerito")

numero = int(dato)
if numero > 0:
    print(f'El número {numero} es mayor que cero')
elif numero ==0:
    print(f'El número {numero} es cero')
else:
    print(f'El número {numero} es menor que cero')
```

```python
cadena = "texto de prueba"
letra = input("Dime una letra")
if letra in cadena:
    print(f'La letra {letra} está en la cadena')
else:
    print(f'La letra {letra} no está en la cadena')
    
```

### 1. Escriba un programa que solicite al usuario que ingrese una cadena. El programa debería mostrar lo siguiente:


(a) El número total de caracteres de la cadena

(b) La cadena repetida 10 veces

(c) El primer carácter de la cadena (recuerde que los índices de cadena empiezan en 0)

(d) Los tres primeros caracteres de la cadena

(e) Los tres últimos caracteres de la cadena

(f) La cadena al revés

(g) El séptimo carácter de la cadena si la cadena es lo suficientemente larga; en caso contrario, un mensaje

(h) La cadena sin el primer y el último carácter

(i) La cadena en mayúsculas

(j) La cadena con cada a reemplazada por una e

(k) La cadena con cada letra reemplazada por un asterisco


```python
cadena = input("Dime una cadena")
print(f'La cadena es "{cadena}"')
```

```python
print(f'A. El número total de caracteres es {(len(cadena))}')
```

```python
for i in range(10):
    print(f'B "{cadena}"')
```

```python
print(f'C. El primer caracter de la cadena es "{(cadena[0])}"')
```

```python
print(f'D. Los tres primeros caracteres de la cadena son "{(cadena[0:3:1])}"')
```

```python
print(f'E. Los tres últimos caracteres de la cadena son "{(cadena[-3:])}"')
```

```python
print(f'F. La cadena al revés es "{(cadena[::-1])}"')
```

```python
if len(cadena) > 6:
    print(f'G. El septimo caracter es  "{(cadena[6])}"')
else:
    print(f'G. La cadena es corta')
```

```python
print(f'H. {cadena[1:-1]}')
```

```python

print(f'I. {cadena.upper()}')
```

```python
print(f'J. {cadena.replace('a','e')}')
```

```python
otra = ''
for letra in cadena:
    otra += '*'
print(f'K. {otra}')

```

# BUCLES

```python
for i in range(10):
    print(i)
```

```python
for i in range(1,25,3):
    print(i)
```

```python
for i in range (25,1,-3):
    print(i)
```

```python
lista = [1,2,3]
otra = [4,5,6]
total = lista + otra
print(total)
```

```python
for n in total:
    if n % 2 ==0:
        print(n)
    else:
        print('-')

```

# Bucles while

```python
contador = 0
while contador < 10:
    print(contador)
    contador += 1

print("Fuera del bucle")
```

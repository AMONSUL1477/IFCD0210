### 1. El código a continuación imprime los números del 1 al 50.

Reescribe el código usando un bucle while para lograr lo mismo.

for i in range(1,51):
    print(i)

```python
i = 1
while i < 51:
    print(i)
    i += 1
```

### (b) Modifica el programa anterior para imprimir cada segundo carácter de la cadena.

### 3. Un buen programa se asegurará de que los datos que ingresan los usuarios sean válidos.

### Escribe un programa que solicite al usuario un peso y lo convierta de kilogramos a libras.
### Siempre que el usuario ingrese un peso menor a 0, el programa le indicará que su entrada no es válida y le solicitará nuevamente que ingrese un peso.
### [Sugerencia: Usa un bucle while, no una sentencia if].

```python
kg = float(input("Dime un peso en kg"))

FACTOR = 2.204623


while kg < 0:
    kg = float(input("Dime un peso en kg"))

print(f'{kg} kilos son {kg*FACTOR} libras')


```

### 5. Escriba un programa que permita al usuario ingresar cualquier número de puntajes de prueba.
### El usuario indica que ha terminado ingresando un número negativo.
### Imprima cuántos puntajes son A (90 o superiores).
### Imprima también el promedio.

```python
puntaje = []
mas_90 = 0
promedio = 0

while True:
    numero = input("Puntuaje")
    if numero < 0:
        break
    puntaje.append(numero)

    if numero >= 90:
        mas_90 += 1

promedio = sum(puntaje) / len(puntaje)

print(f'Ma de 90: {mas_90}')
print(f'Promedio {promedio}')


```

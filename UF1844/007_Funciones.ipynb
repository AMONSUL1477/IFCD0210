```python
def mostrar_config(**kwargs):
    for clave, valor in kwargs.items():
        print(f"{clave}: {valor}")

mostrar_config(usr='teo',ip='1,2,3', puerto='8080')
```

```python
d = {
    'usr' : 'teo',
    'ip' : '1.2.3.4',
    'puerto' : '8080'
}

mostrar_config(**d)
```

```python
def log(mensaje,*args,**kwargs):
    nivel = kwargs.get('nivel', 'DEBUG')

    prefijo = kwargs.get('prefijo', '')

    

    msj_formato = f'[{nivel}]:{prefijo} - {mensaje}'

    return msj_formato

log('Error', nivel = 'DEBUG')
    
```

```python
def hhtp_get(url,**kwargs):
    timeout = kwargs.get('timeout',5)
    headers = kwargs.get('headers',{})
    params = kwargs.get('params', {})

    return{
    
    'url': url,
    'timeout' : timeout,
    'headers' : headers,
    'params' : params
    }


hhtp_get(
    'http://google.es',
    timeout = 10,
    params = {'page':2}
)
```
